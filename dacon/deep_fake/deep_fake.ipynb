#### 이미지 파일 처리

```python
transform = T.Compose([
    T.Resize((224, 224)),
    T.ToTensor(),
    T.Normalize(mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225]),
])

image = Image.open('test_data/TEST_001.jpg').convert('RGB')
x = transform(image)
print(x.shape)
```

#### 동영상 파일 처리

- 동영상은 프레임을 뽑아서 이미지 처럼 처리합니다.

```python
def extract_frames(video_path, max_frames=10):
    cap = cv2.VideoCapture(video_path)
    frames = []
    count = 0
    while cap.isOpened() and len(frames) < max_frames:
        ret, frame = cap.read()
        if not ret:
            break

        if count % 10 == 0:  # 간격 샘플링
            frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            frames.append(frame)
        count += 1

    cap.release()
    return frames
```

```python
import cv2

# 1. 비디오 파일 열기 (경로 지정)
# 예시: cap = cv2.VideoCapture('my_video.mp4')
cap = cv2.VideoCapture('train_data/real/original/000.mp4') # 'video.mp4'를 실제 파일 경로로 변경하세요.

# 2. 비디오가 제대로 열렸는지 확인
if not cap.isOpened():
    print("비디오 파일을 열 수 없습니다.")
else:
    # 3. 비디오의 각 프레임을 읽어와 화면에 표시
    while True:
        # ret: 프레임 읽기 성공 여부, frame: 읽어온 프레임 (이미지)
        ret, frame = cap.read()

        # 파일 끝에 도달했거나 프레임을 읽지 못하면 루프 종료
        if not ret:
            print("비디오 재생 완료.")
            break

        # 읽어온 프레임을 화면에 출력
        cv2.imshow('Video Playback', frame)

        # cv2.waitKey(1): 1ms(밀리초) 동안 키 입력을 기다림. 값이 클수록 느려짐.
        # 'q' 키가 눌리면 종료
        if cv2.waitKey(25) & 0xFF == ord('q'):
            break

# 4. 자원 해제
cap.release() # 비디오 캡처 객체 해제
cv2.destroyAllWindows() # 열려있는 모든 창 닫기
```
